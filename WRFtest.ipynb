# WPS Configuration

```python
# WPS Setup
# Note - Namelist.wps is not a normal text file, will download and open weird.
# Link WPS GFS Files
%%bash
export LD_LIBRARY_PATH="/content/wrf_dependencies/grib2/lib"

cd /content/WPS
./link_grib.csh /content/drive/MyDrive/colab/gfs/gfs*
ln -sf ungrib/Variable_Tables/Vtable.GFS Vtable

# Make sure WPS_GEOG exists and is linked under .../colab/WPS_GEOG
# WPS_GEOG can be acquired from NCAR - http://www2.mmm.ucar.edu/wrf/users/download/get_sources_wps_geog.html. Needs to be
# untarred, expanding to about 30gb - recommend to save to your drive to save time.

export wpsfile=namelist.wps

GEOG_DIR="/content/drive/MyDrive/colab/WPS_GEOG"

if [[ -d "$GEOG_DIR" ]]; then
    echo "Found WPS geogrid directory: $GEOG_DIR"
    sed -i "s|^[[:space:]]*geog_data_path.*| geog_data_path = '$GEOG_DIR'|" "$wpsfile"
else
    echo "ERROR: WPS geogrid directory not found: $GEOG_DIR"
    exit 1
fi

export nml="/content/WPS/namelist.wps"

# Edit WPS namelist parameters - easier here than in the run script since they dont change.
sed -i "s/^[[:space:]]*map_proj.*/ map_proj = 'mercator', /" "$nml"
sed -i "s/^[[:space:]]*ref_lat.*/ ref_lat = -19.0, /" "$nml"
sed -i "s/^[[:space:]]*ref_lon.*/ ref_lon = 58.0, /" "$nml"
sed -i "s/^[[:space:]]*truelat1.*/ truelat1 = -21.0, /" "$nml"
sed -i "s/^[[:space:]]*truelat2.*/ truelat2 = -19.0, /" "$nml"
sed -i "s/^[[:space:]]*stand_lon.*/ stand_lon = 63.0, /" "$nml"
sed -i "s/^[[:space:]]*out_format.*/ out_format = 'WPS', /" "$nml"
sed -i "s/^[[:space:]]*prefix.*/ prefix = 'GFS', /" "$nml"
sed -i "s/^[[:space:]]*fg_name.*/ fg_name = 'GFS', /" "$nml"
sed -i "s/^[[:space:]]*map_proj.*/ map_proj = 'lambert', /" "$nml"

if ! grep -q "io_form_metgrid" "$nml"; then
    echo "Adding io_form_metgrid = 2 to &metgrid section..."

    sed -i '/&metgrid/,/\// {
        /\//! {
            s|/| io_form_metgrid = 2,\n/|
        }
    }' "$nml"
else
    echo "io_form_metgrid already present."
fi



```

```python
%%bash
# Sometimes WPS does not compile with a couple necessary lines. This is a check for that.
export wpsfile="/content/WPS/namelist.wps"

awk '
BEGIN { in_metgrid=0; seen=0 }

/^[[:space:]]*&metgrid([[:space:]]*|[[:space:]]*!.*)$/ {
  in_metgrid=1; seen=0
  print
  next
}

in_metgrid {
  # If the line exists in &metgrid, replace it
  if ($0 ~ /^[[:space:]]*io_form_metgrid[[:space:]]*=/) {
    print " io_form_metgrid = 2,"
    seen=1
    next
  }

  # End of &metgrid block: insert if missing, then print the closing /
  if ($0 ~ /^[[:space:]]*\/[[:space:]]*$/) {
    if (!seen) print " io_form_metgrid = 2,"
    print
    in_metgrid=0
    next
  }

  print
  next
}

{ print }
' "$wpsfile" > "${wpsfile}.tmp" && mv "${wpsfile}.tmp" "$wpsfile"

# Verify
echo ">>> &metgrid block now contains:"
awk '/^[[:space:]]*&metgrid/{flag=1} flag{print} /^[[:space:]]*\/[[:space:]]*$/{if(flag){flag=0; exit}}' "$wpsfile"
```

- My script is set up for manually creating trials, i.e. going into the script and simply changing the assimilated obs and changing the "trial" variable. The script will handle the file structures and everything else.

- You can save compute time by using the same coarse run for multiple trials. Then, simply comment out line 926 "run_wrf_coarse_96h".

Within cell 16, change the directory path to where your WPS_GEOG file is stored. You also need to adjust the path under "WPS Configuration" to lead ./link_csh to your gfs files.
- In my script, I have the variables I change frequently set up (like what obs are assimilated). More changing namelist variables can be added (such as in WRF or WRFDA) but the basics for further research are there. Just follow the patterns in the update_namelist functions to add more.
- In my script, I hold copies of the the namelist with most of the stuff filled out within my drive, which I just copy over then edit for things like time after removing the default generated namelist. Was easier than having to change every little thing every time.
- DO NOT trust AI with the namelist. In fact, it knows extremely little about WRF or NWP; use with EXTREME caution. To use it effectively, give it very small tasks and give it as much context as possible.

The exception here is the WPS_GEOG directory and the ./link_csh to GFS function, which for some reason doesnt throw a fit (most likely because its just a set of raw text files - but so is CRTM so who knows). Couldn't tell you why it be like this but if you're getting weird errors its probably this.
